```python
#import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Create a dataframe
df=pd.read_csv("orders.csv")
df.head()
```

```python
#Basic NumPy Analysis
print(np.mean(df['Sales']))
print(np.median(df['Sales']))
print(np.std(df['Sales']))
print(np.max(df['Sales']))
print(np.min(df['Sales']))
```

```python
#Data Cleaning with pandas
#Checking missing values
df.isnull().sum()
```

```python
#Remove duplicates
df.drop_duplicates(inplace=True)
```

```python
#Convert dates
df['Order_Date']=pd.to_datetime(df['Order_Date'])
```

```python
df.head(2)
```

```python
df.columns
```

```python
df['Revenue']=df['Quantity']+df['Unit_Price']
```

```python
df.head(2)
```

```python
df['Year']=df['Order_Date'].dt.year
df['Month']=df['Order_Date'].dt.month
df['Month_Name']=df['Order_Date'].dt.month_name()
```

```python
df.head(2)
```

```python
#Business Analysis  [E-Commerce Sales Analysis using pandas]
#Q. 1. What is total revenue ?
print(df['Revenue'].sum())
```

```python
#Q. 2 What is total profit ?
print(df['Profit'].sum())
```

```python
#Q.3 Which product generates highest sales ?
print(df.groupby('Product')['Revenue'].sum().sort_values(ascending=False))
```
